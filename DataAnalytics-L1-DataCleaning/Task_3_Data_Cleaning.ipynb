```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Load the Titanic dataset
df = pd.read_csv("Titanic-Dataset.csv")

# Display first 5 rows
df.head()
```

```python
# Check the number of rows and columns
print("Dataset Shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
df.describe()
```

```python
# Create a copy of the original dataset
df_before = df.copy()

print("Original dataset backup created successfully!")
```

```python
# Create data quality report

quality_report = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100,
    "Unique Values": df.nunique()
})

# Display report
quality_report
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
print("Age Range:", df["Age"].min(), "to", df["Age"].max())
print("Fare Range:", df["Fare"].min(), "to", df["Fare"].max())
print("Negative Age Values:", (df["Age"] < 0).sum())
print("Negative Fare Values:", (df["Fare"] < 0).sum())
```

```python
print(df.isnull().sum())
```

```python
print("Missing Age Values:", df["Age"].isnull().sum())
```

```python
print("Cabin" in df.columns)
```

```python
print(df["Embarked"].mode())
```

```python
print("Missing Embarked Values:", df["Embarked"].isnull().sum())
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Rows After Duplicate Removal:", df.shape[0])
print("Duplicate Rows Remaining:", df.duplicated().sum())
```

```python
df["Sex"] = df["Sex"].str.strip().str.title()
df["Embarked"] = df["Embarked"].str.strip().str.upper()

print("Sex Values:", df["Sex"].unique())
print("Embarked Values:", df["Embarked"].unique())
```

```python
df["Name"] = df["Name"].str.strip()
df["Ticket"] = df["Ticket"].str.strip()

print("Extra spaces removed successfully!")
print(df[["Name", "Ticket"]].head())
```

```python
Q1 = df[["Age", "Fare"]].quantile(0.25)
Q3 = df[["Age", "Fare"]].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = ((df[["Age", "Fare"]] < lower_limit) |
            (df[["Age", "Fare"]] > upper_limit)).sum()

print("Outliers in Age:", outliers["Age"])
print("Outliers in Fare:", outliers["Fare"])
```

```python
print(df.dtypes)
```

```python
df["PassengerId"] = df["PassengerId"].astype(str)

print(df.dtypes)
```

```python
print("BEFORE CLEANING")
print("Rows:", df_before.shape[0])
print("Missing Values:", df_before.isnull().sum().sum())
print("Duplicate Rows:", df_before.duplicated().sum())

print("\nAFTER CLEANING")
print("Rows:", df.shape[0])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.to_csv("Titanic_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
